## Introduction to web scraping 

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 
```

```python
car_data = []
for page in range(1,4):
    url = f'https://jiji.co.ke/cars?filter_attr_1_make=Toyota&slug=cars&init_page=false&page={page}'
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    cars = soup.find_all('div', class_='masonry-item')
    

    for car in cars:
        price = car.find("div",class_="b-list-advert-base__data__price").text
        name = car.find("div",class_="b-list-advert-base__data__title").text
        desc =  car.find("div",class_="b-list-advert-base__description-text").text
        car_data.append({"name":name,"price":price,"desc":desc})
    
df = pd.DataFrame(car_data)
df.to_csv("car_data.csv")
df.describe()
```
